```python
%sql
DROP TABLE IF EXISTS la_lakehouse.gold.fact_permits;
```

```python
%sql
CREATE OR REPLACE TABLE  la_lakehouse.gold.fact_permits (
    permit_nbr STRING PRIMARY KEY,
    valuation INTEGER,
    square_footage FLOAT,
    du_changed INTEGER,
    zip_code INTEGER,
    adu_changed STRING,
    height FLOAT,
    primary_address STRING,
    work_desc STRING,
    status_desc STRING,
    status_date TIMESTAMP,
    construction STRING,
    lat FLOAT,
    lon FLOAT,
    type_lat_lon STRING,
    geolocation BINARY,
    pin_nbr STRING,
    square_footage_multi BOOLEAN,
    zone_is_multi BOOLEAN,
    ct_multi BOOLEAN,
    issue_date_key BIGINT,
    cofo_date_key BIGINT,
    submitted_date_key BIGINT,

    zone_key BIGINT REFERENCES la_lakehouse.gold.dim_zone(zone_key),

    CONSTRAINT fk_issue_date FOREIGN KEY (issue_date_key) REFERENCES la_lakehouse.gold.dim_date(date_key),

    CONSTRAINT fk_cofo_date FOREIGN KEY (cofo_date_key) REFERENCES la_lakehouse.gold.dim_date(date_key),

    CONSTRAINT fk_submitted_date FOREIGN KEY (submitted_date_key)  REFERENCES la_lakehouse.gold.dim_date(date_key),

    ct_key BIGINT REFERENCES la_lakehouse.gold.dim_census_tract(ct_key),

    district_key BIGINT REFERENCES la_lakehouse.gold.dim_district(district_key),

    apc_key BIGINT REFERENCES la_lakehouse.gold.dim_area_planning_commision(apc_key),

    cpa_key BIGINT REFERENCES la_lakehouse.gold.dim_community_plan_area(cpa_key),

    cnc_key BIGINT REFERENCES la_lakehouse.gold.dim_certified_neighborhood_council(cnc_key),

    hl_key BIGINT REFERENCES la_lakehouse.gold.dim_hillside_ordinance_area(hl_key),

    permit_type_key BIGINT REFERENCES la_lakehouse.gold.dim_permit_type(permit_type_key),

    use_type_key BIGINT REFERENCES la_lakehouse.gold.dim_use_type(use_type_key)


)
```
