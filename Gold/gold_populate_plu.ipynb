```python
%sql
MERGE INTO la_lakehouse.gold.fact_plum_items AS target 
USING (
    SELECT
    ps.document_id,
    ps.extraction_failed,
    ps.ingested_at,
    ps.meeting_date,
    ps.meeting_id,
    ps.meeting_title,
    ps.template_id,
    ps.template_name,
    ps.item_position,
    ps.item_number,
    ps.council_file_no,
    ps.council_district,
    ps.applicant,
    ps.representative,
    ps.case_no,
    ps.environmental_no,
    ps.fiscal_impact_statement,
    ps.financial_policies_statement,
    ps.continued_from_date,
    ps.is_continued,
    ps.community_impact_reasons,
    ps.community_impact_submitted,
    dd.district_key
    FROM la_lakehouse.silver.plum_items AS ps
    LEFT JOIN la_lakehouse.gold.dim_district AS dd
    ON ps.council_district = dd.cd
) AS source
ON target.document_id = source.document_id AND target.item_number = source.item_number
WHEN NOT MATCHED THEN 
    INSERT(
    document_id,
    extraction_failed,
    ingested_at,
    meeting_date,
    meeting_id,
    meeting_title,
    template_id,
    template_name,
    item_position,
    item_number,
    council_file_no,
    council_district,
    applicant,
    representative,
    case_no,
    environmental_no,
    fiscal_impact_statement,
    financial_policies_statement,
    continued_from_date,
    is_continued,
    community_impact_reasons,
    community_impact_submitted,
    district_key
    )
    VALUES(
    source.document_id,
    source.extraction_failed,
    source.ingested_at,
    source.meeting_date,
    source.meeting_id,
    source.meeting_title,
    source.template_id,
    source.template_name,
    source.item_position,
    source.item_number,
    source.council_file_no,
    source.council_district,
    source.applicant,
    source.representative,
    source.case_no,
    source.environmental_no,
    source.fiscal_impact_statement,
    source.financial_policies_statement,
    source.continued_from_date,
    source.is_continued,
    source.community_impact_reasons,
    source.community_impact_submitted,
    source.district_key
    );
```
