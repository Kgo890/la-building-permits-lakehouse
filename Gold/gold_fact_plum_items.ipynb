```python
%sql
CREATE OR REPLACE TABLE la_lakehouse.gold.fact_plum_items (
    document_id INTEGER,
    extraction_failed BOOLEAN,
    ingested_at TIMESTAMP,
    meeting_date DATE,
    meeting_id INTEGER,
    meeting_title STRING,
    template_id INTEGER,
    template_name STRING,
    item_position INTEGER,
    item_number INTEGER,
    council_file_no STRING,
    council_district INTEGER,
    applicant STRING,
    representative STRING,
    case_no STRING,
    environmental_no STRING,
    fiscal_impact_statement STRING,
    financial_policies_statement STRING,
    continued_from_date DATE,
    is_continued BOOLEAN,
    community_impact_reasons STRING,
    community_impact_submitted BOOLEAN,

    district_key BIGINT REFERENCES la_lakehouse.gold.dim_district(district_key),
    PRIMARY KEY (document_id, item_number)
)
```
